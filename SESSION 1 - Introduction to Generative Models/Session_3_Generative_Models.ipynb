### 2.
**Create a table comparing discriminative models and generative models across at least 4 aspects (definition, main use, example task, example model name).**

---
| Aspect | Discriminative Models | Generative Models |
| :--- | :--- | :--- |
| **Definition** | Learns the boundary between different classes (P(Y\|X)). | Learns the underlying distribution of the data itself (P(X)). |
| **Main Use** | Classification, categorization, and prediction based on existing data. | Creation, synthesis, and simulation of brand-new data. |
| **Example Task** | Identifying whether a Zomato review is Positive or Negative. | Writing a completely new, fake Zomato review. |
| **Example Model Name** | Logistic Regression, Support Vector Machines (SVM). | Generative Adversarial Networks (GANs), Large Language Models (LLMs). |

### 4.
**Pick one generative AI application (text generation, image synthesis, music/audio generation, or code generation) and describe a new feature you would add to an app you use, explaining how generative AI would power it.**

---
**Gen AI Application:** Image Synthesis
**App:** Zomato
**New Feature:** *'Crave-Visualizer'*

**How GenAI powers it:** 
Often, smaller restaurants don't have the budget to take professional photos for every single item on their menu. Using an Image Synthesis model (like Stable Diffusion), Zomato could automatically read the text description of the dish (e.g., 'Spicy paneer tikka with green chutney') and instantly generate an extremely realistic, mouth-watering image of exactly what that food looks like, so customers never have to order 'blind' again.
