# What is GenAI?


### 1.
**Write a short paragraph in your own words explaining what Generative AI (GenAI) is, and give two real-world examples of GenAI applications you have seen or used (such as ChatGPT, DALL-E, or Instagram filters).**

---
Generative AI (GenAI) is a branch of artificial intelligence focused on creating brand new content—like text, images, audio, or code—based on patterns it has learned from massive amounts of training data, rather than just analyzing or categorizing existing information. Two real-world examples I use are **ChatGPT** (which generates new conversational text and explanations) and **Midjourney/DALL-E** (which generates entirely new images from text descriptions).


### 2.
**Create a table comparing GenAI with traditional AI by listing at least three key differences in how they work and the types of outputs they produce.**

---
| Feature | Traditional AI (Analytical) | Generative AI (GenAI) |
| :--- | :--- | :--- |
| **Core Function** | Analyzes data, predicts outcomes, or categorizes information (e.g., classifying emails as spam). | Creates completely new, original content (e.g., writing a new email). |
| **Output Format** | Fixed outputs like numbers, classes, labels, or probabilities. | Creative outputs like paragraphs of text, images, videos, or code. |
| **Primary Goal** | To understand, sort, or find patterns in existing data. | To synthesize and generate net-new data that mimics human creativity. |


### 3.
**List 5 popular apps or features you use daily (like Instagram Reels, Spotify recommendations, Zomato chatbots, YouTube video summaries, or Flipkart product images) and identify which ones likely use GenAI. For each, briefly explain how GenAI might be working behind the scenes.**

---
1. **Instagram (AI Filters / Avatars)** - Uses GenAI. It generates new pixels in real-time to alter your face or create a stylized avatar based on your facial movements.
2. **Spotify Recommendations** - *Traditional AI*. It analyzes your listening history and predicts what existing songs you will like (pattern matching, not generating new songs).
3. **Zomato Support Chatbots** - Uses GenAI. Modern support bots use Large Language Models (LLMs) to understand context and generate natural, conversational replies to solve complex refund issues on the fly instead of relying on rigid, pre-written templates.
4. **YouTube Video Summaries** - Uses GenAI. It takes the video transcript as input and generates a completely new, condensed paragraph summarizing the key points of the video.
5. **Flipkart Product Search** - *Traditional AI*. It uses traditional search algorithms and pattern matching to retrieve existing products from a database based on your typed keywords.
