### 1.
**Write a Python function called `simple_generator()` that takes a random vector as input and returns a 28x28 array of grayscale pixel values between 0 and 1, simulating a fake handwritten digit image like those used in GAN demos.**


```python
import numpy as np

def simple_generator(random_vector):
    np.random.seed(int(np.sum(random_vector) * 1000) % 100000)
    return np.random.rand(28, 28)
```

### 2.
**Create a Python class Discriminator with a method discriminate(image) that takes a 28x28 image array and returns a probability score between 0 and 1 indicating if the image is 'real' or 'fake'.**


```python
class Discriminator:
    def discriminate(self, image):
        mean_val = np.mean(image)
        probability_real = max(0.0, 1.0 - (mean_val * 1.5))
        return probability_real
```

### 3.
**Simulate one cycle of the adversarial training loop: generate a fake image using your simple_generator(), get a score from your Discriminator, and print whether the Discriminator correctly identifies the fake image.**


```python
latent_vector = np.random.rand(100)
fake_image = simple_generator(latent_vector)
```

```python
disc = Discriminator()
score = disc.discriminate(fake_image)
```

```python
threshold = 0.5
is_identified_as_fake = score < threshold
```

```python
print(f"Discriminator Score (Probability of being real): {score}")
print(f"Did Discriminator correctly identify the fake image? {is_identified_as_fake}")
```
